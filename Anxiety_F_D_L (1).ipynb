```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, label_binarize
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix,
    roc_curve, auc
)
from sklearn.utils import resample
from xgboost import XGBClassifier

# === Define files by domain ===
fd_files = [
    'FD HF Percentage.csv', 'FD HF Power.csv', 'FD LF Percentage.csv',
    'FD LF Power.csv', 'FD VLF Power.csv', 'FD VLF Percentage.csv',
    'FD LF HF Ratio.csv'
]

td_files = [
    'TD HR Mean.csv', 'TD HR Std.csv', 'TD NN 50.csv', 'TD PNN 50.csv',
    'TD RMSSD .csv', 'TD RR Mean.csv', 'TD SDNN.csv', 'TD TINN.csv',
    'TD Triangular Index.csv'
]

la_files = [
    'LA Ratio.csv', 'LA SD1.csv', 'LA SD2.csv'
]

all_files = fd_files + td_files + la_files
phases = ['Task 1', 'Task 2', 'Task 3', 'Rest 1', 'Rest 2', 'Rest 3']

# === Load and prepare dataset ===
all_dfs = []
for file in all_files:
    df = pd.read_csv(file)
    df['Subject'] = df.index
    feature_name = file.replace('.csv', '').strip()
    melted = df.melt(id_vars=['Subject', 'Class'], var_name='Phase', value_name='Value')
    melted = melted[melted['Phase'].isin(phases)]
    melted['Feature'] = feature_name
    all_dfs.append(melted)

df_long = pd.concat(all_dfs, ignore_index=True)
df_wide = df_long.pivot_table(index=['Subject', 'Phase', 'Class'], columns='Feature', values='Value').reset_index()
df_wide['Label'] = df_wide['Phase'] + ' - ' + df_wide['Class'].map({0: 'Kaapi', 1: 'Weightless'})

# Encode labels
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(df_wide['Label'])
class_names = label_encoder.classes_
y_bin = label_binarize(y, classes=np.unique(y))

# Separate domains
fd_cols = [f.replace('.csv', '').strip() for f in fd_files]
td_cols = [f.replace('.csv', '').strip() for f in td_files]
la_cols = [f.replace('.csv', '').strip() for f in la_files]
all_cols = df_wide.columns.difference(['Subject', 'Phase', 'Class', 'Label'])

# Normalize all
X_all = df_wide[all_cols].fillna(df_wide[all_cols].mean())
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_all)

# Bootstrap and split
X_resampled, y_resampled = resample(X_scaled, y, replace=True, n_samples=len(y), random_state=42)
y_bin_resampled = label_binarize(y_resampled, classes=np.unique(y))

X_train, X_test, y_train, y_test, y_bin_train, y_bin_test = train_test_split(
    X_resampled, y_resampled, y_bin_resampled, stratify=y_resampled, test_size=0.2, random_state=42
)

# Models
models = {
    "Random Forest": RandomForestClassifier(random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "SVM": CalibratedClassifierCV(SVC(kernel='rbf', random_state=42), cv=3),
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "XGBoost": XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=42)
}

# === Overall Classification Metrics ===
print("\n=== Overall Model Metrics (All Features) ===")
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    print(f"\n{name} Accuracy: {acc:.4f}")
    print("Classification Report:")
    print(classification_report(y_test, y_pred, target_names=class_names))
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names)
    plt.title(f"{name} - Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# === Feature Ablation (Random Forest) ===
print("\n=== Feature Ablation (Random Forest) ===")
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
base_acc = accuracy_score(y_test, rf.predict(X_test))
print(f"Base Accuracy: {base_acc:.4f}")

for i, col in enumerate(all_cols):
    X_train_ab = np.delete(X_train, i, axis=1)
    X_test_ab = np.delete(X_test, i, axis=1)
    rf_ab = RandomForestClassifier(random_state=42)
    rf_ab.fit(X_train_ab, y_train)
    acc_ab = accuracy_score(y_test, rf_ab.predict(X_test_ab))
    print(f"Removed {col}: Accuracy = {acc_ab:.4f} | Δ = {base_acc - acc_ab:.4f}")

# === ROC Curve Comparisons by Domain ===
domain_sets = {
    "FD": fd_cols,
    "TD": td_cols,
    "LA": la_cols,
    "ALL": list(all_cols)
}

print("\n=== Macro-Averaged ROC Curves by Domain ===")
for domain, features in domain_sets.items():
    print(f"\n>> {domain} Features")
    X_dom = df_wide[features].fillna(df_wide[features].mean())
    X_dom_scaled = StandardScaler().fit_transform(X_dom)
    X_res, y_res = resample(X_dom_scaled, y, replace=True, n_samples=len(y), random_state=42)
    y_bin_res = label_binarize(y_res, classes=np.unique(y))
    X_train_d, X_test_d, y_train_d, y_test_d, y_bin_train_d, y_bin_test_d = train_test_split(
        X_res, y_res, y_bin_res, stratify=y_res, test_size=0.2, random_state=42)

    plt.figure(figsize=(8, 6))
    for name, model in models.items():
        model.fit(X_train_d, y_train_d)
        y_score = model.predict_proba(X_test_d)

        all_fpr = np.linspace(0, 1, 100)
        mean_tpr = np.zeros_like(all_fpr)

        for i in range(len(class_names)):
            fpr, tpr, _ = roc_curve(y_bin_test_d[:, i], y_score[:, i])
            mean_tpr += np.interp(all_fpr, fpr, tpr)
        mean_tpr /= len(class_names)
        mean_tpr[-1] = 1.0
        macro_auc = auc(all_fpr, mean_tpr)

        plt.plot(all_fpr, mean_tpr, lw=2, label=f"{name} (AUC = {macro_auc:.2f})")

    plt.plot([0, 1], [0, 1], 'k--')
    plt.title(f"{domain} - Macro-Averaged ROC")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend(loc='lower right')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.utils import resample

# === Load your df_wide, y, and class_names from your main script ===

# Define feature groups
fd_cols = [
    'FD HF Percentage', 'FD HF Power', 'FD LF Percentage',
    'FD LF Power', 'FD VLF Power', 'FD VLF Percentage', 'FD LF HF Ratio'
]

td_cols = [
    'TD HR Mean', 'TD HR Std', 'TD NN 50', 'TD PNN 50',
    'TD RMSSD', 'TD RR Mean', 'TD SDNN', 'TD TINN',
    'TD Triangular Index'
]

la_cols = ['LA Ratio', 'LA SD1', 'LA SD2']

domain_sets = {
    'FD': fd_cols,
    'TD': td_cols,
    'LA': la_cols,
    'ALL': list(fd_cols + td_cols + la_cols)
}

# === Function to compute feature importance ===
def compute_tree_importance(domain, features):
    X = df_wide[features].fillna(df_wide[features].mean())
    X = StandardScaler().fit_transform(X)
    X_resampled, y_resampled = resample(X, y, replace=True, n_samples=len(y), random_state=42)

    model = RandomForestClassifier(random_state=42)
    model.fit(X_resampled, y_resampled)

    importances = model.feature_importances_
    fi_df = pd.DataFrame({'Feature': features, 'Importance': importances})
    fi_df = fi_df.sort_values(by='Importance', ascending=False)

    # Plot
    plt.figure(figsize=(10, 6))
    plt.barh(fi_df['Feature'], fi_df['Importance'], color='teal')
    plt.gca().invert_yaxis()
    plt.title(f'Feature Importance ({domain}) - Random Forest')
    plt.xlabel('Importance')
    plt.tight_layout()
    plt.show()

    return fi_df

# === Run for each domain ===
for domain, features in domain_sets.items():
    print(f"\n=== {domain} Feature Importance ===")
    compute_tree_importance(domain, features)
```
